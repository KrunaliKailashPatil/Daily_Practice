```python
# Question 1. Calculate the Product of Digits
def product_digit(number):
    product = 1

    while number > 0:
        digit = number % 10
        product = product * digit
        number = number // 10

    return product
print(product_digit(234))
```

```python
# Question 2. Calculate Overtime Pay
def overtime(hours):
    if hours > 40:
        extra_pay = hours - 40
        return extra_pay * 200
    else:
        return "No extra pay"
print(overtime(45))
print(overtime(38))
```

```python
# Question 3. Count Uppercase and Lowercase Letters
def count(letter):
    upper = 0
    lower = 0

    for text in letter:
        if text.isupper():
            upper = upper + 1
        elif text.islower():
            lower = lower + 1
    print("Upper:",upper)
    print("Lower:",lower)
    
print(count(["A","B","c","D","e"]))
```

```python
# Question 4. Find Common Elements in Two Lists
list1=[10,20,30,40,50]
list2=[40,50,60,70,80]

def common_element(list1,list2):
    result = []

    for item in list1:
        if item in list2 and item not in result:
            result.append(item)

    return result
print("Common element:",common_element(list1,list2))
```

```python
# Question 6. Find the Shortest Word
def shortest_list(text):
    short = text[0]

    for texts in text:
        if len(texts) < len(short):
            short = texts
    return short

print(shortest_list(["python","sql","java","kotlin"]))
```

```python
# Question 7. Function for prime number 
def prime(number):
    for num in range(2,number):
        if number %  num == 0:
            return False
        else:
            return True

print(prime(10))
```

```python
# Question 8. Calculate Average Order Value
def order_value(orders):
    total = sum(orders)
    count = len(orders)
    return total/count

print("Average order value:",(order_value([10,20,30,40,50])))
```
