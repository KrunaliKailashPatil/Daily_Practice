```python
# Question 1. Calculate the Area of a Rectangle
def calculate(length,width):
    result = length * width
    return result

print(calculate(100,89))
```

```python
# Question 2. Calculate Electricity Bill
def electricity_bill(unit):
    bill = unit * 6
    return bill

print(electricity_bill(100))
```

```python
# Question 3. Calculate Percentage
def calculate_percentage(marks):
    total = sum(marks)
    percenatge = (total / 500) * 100
    return percenatge

print(calculate_percentage([89,98,77,88,66]))
```

```python
# Question 4. Separate Even and Odd Numbers
def seprate(num):
    even = []
    odd = []

    for num in num:
        if num % 2 == 0:
            even.append(num)
        else:
            odd.append(num)

    print("Even =",even)
    print("Odd =",odd)

print(seprate([12,4,3,90,0,43]))
```

```python
# Question 5. Find Numbers Greater Than the Average
def greater_avg(number):
    avg = sum(number) / len(number)
    result = []
    for num in number:
        if num > avg:
            result.append(num)
    return result

print(greater_avg([10,20,20,30]))
```

```python
# Question 6. Find the Longest Word
def longest_word(words):
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

print(longest_word(["SQL","PYTHON","JAVA"]))
```

```python
# Question 7. Count Numbers Above a Given Limit
def count_number(number,limit):
    count = 0
    for num in number:
        if num > limit:
           count = count + 1
    return count

print(count_number([12,43,54,65,23,23,76,87],50)) 
```

```python
# Question 8. Remove Duplicate Values
def remove_duplicate(number):
    result = []
    for num in number:
        if num not in result:
            result.append(num)
    return result

print(remove_duplicate([120,49,120,49,33,48,99,99]))
```

```python
# Question 9. Categorize Employee Salaries
def category(salaries):
    high = []
    medium = []
    low =[]

    for salary in salaries:
        if salary >= 100000:
            high.append(salary)
        elif salary >= 50000:
            medium.append(salary)
        else:
            low.append(salary)
    print("High Salary:",high)
    print("Medium Salary:",medium)
    print("Low Salary:",low)

print(category([20000,50000,70000,100000,90000]))
```

```python
# Question 10. Find the Most Frequent Number
def Frequent(number):
    common = number[0]
    highest = number.count(common)

    for num in number:
        count = number.count(num)

        if count > highest:
            highest = count
            common = count
    return common

print(Frequent([10,20,30,40,10,20,10]))
```
