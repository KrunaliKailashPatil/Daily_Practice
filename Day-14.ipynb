```python
# Question 1: Find the Sum of Numbers at Even Indexes
num = [10,20,30,40,50]
total = 0
for i in range(0,len(num),2):
    total += num[i]
print(total)
```

```python
# Question 2: Replace Negative Numbers with Zero
num = [10,20,30,40,50,-10,-20,-50]
for i in range(len(num)):
    if num[i] < 0:
        num[i] = 0
print(num)
```

```python
# Question 3: Find All Numbers Greater Than the Average
num = [10,20,30,40,50]
avg = sum(num)/len(num)
result = []
for i in num:
    if i > avg:
        result.append(num)
print(avg) 
print(result)
```

```python
# Question 5: Remove All Occurrences of a Given Value
numbers = [10, 20, 30, 20]
result = []
for num in numbers:
    if num != 20:
        result.append(num)
print(result)
```

```python
# Question 6: Find Common Unique Elements in Two Lists
list1=[10,20,30,40,50]
list2=[30,40,50,60,70]
common = []
for i in list1:
    if i in list2 and i not in common:
        common.append(i)
print(common)
```

```python
# Question 7: Separate Even and Odd Numbers
num = [10,12,23,34,45,56,67,78,89,90,9,87,76,65]
even = []
odd = []
for i in num:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even:",even)
print("Odd:",odd)    
```

```python
# Question 8: Find the Difference Between Two Lists
list1=[10,20,30,40,50]
list2=[30,40,10,60,70]
result = []
for num in list1:
    if num not in list2:
        result.append(num)
print(result)
```

```python
# Question 10: Move All Elements Below 50 to a New List
num = [10,12,23,34,45,56,67,78,89,90,9,87,76,65]
below = []
other = []
for i in num:
    if i < 50:
        below.append(i)
    else:
        other.append(i)
print("Below 50:",below)
print("Other:",other)
```
