```python
# Question 1. Calculate Simple Interest
def simple_intrest(p,r,t):
    intrest = (p*r*t)/100
    return intrest

print(simple_intrest(5000,4,2))
```

```python
# Question 2. Convert Celsius to Fahrenheit
def celsius_fehrenheit(c):
    f = (c * 9 / 5) + 32
    return f
print(celsius_fehrenheit(25))
```

```python
# Question 3. Calculate Employee Bonus
def calculate_bonus(salary):
    bonus = (salary * 10) / 100
    return bonus

print(int(calculate_bonus(30000)))
```

```python
# Question 4. Count Vowels in a String
def count_vowels(word):
    count = 0

    for ch in word.lower():
        if ch in "aeiou":
            count = count + 1
    return count

print(count_vowels("Apple"))
```

```python
# Question 5. Find the Average of a List
def avg_list(num):
    total = sum(num)
    avg=total/len(num)
    return avg

print(avg_list([10,20,30,40,50]))
```

```python
# Question 6. Check Whether a String Is a Palindrome
def palidrome(word):
    return word == word[::-1]

print(palidrome("python"))
print(palidrome("madam"))
```

```python
# Question 7. Find the Second-Largest Unique Number
def second_largest(num):
    unique_num = sorted(set(num))
    return unique_num[-2]

print(second_largest([12,30,4,54,34,87]))
```

```python
# Question 8. Calculate Discounted Price
def discount(price,discount):
    final_price = (price*discount)/100
    return price - final_price

print(discount(2000,15))
```

```python
# Question 9. Count Positive, Negative, and Zero Values
def count_value(number):
    positive = []
    negative = []
    zero = []

    for num in number:
        if num > 0:
            positive.append(num)
        elif num < 0:
            negative.append(num)
        else:
            zero.append(num)

    print("Postive =",positive)
    print("Negative =",negative)
    print("Zero =",zero)

print(count_value([23,43,-1,0,43,-3,78,0]))
```

```python
# Question 10. Calculate Total Sales Using a Function
def total_sales(sales):
    total = 0

    for amount in sales:
        total = total + amount

    return total

print(total_sales([11200,1500,800,2000]))
```
