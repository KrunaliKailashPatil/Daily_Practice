```python
# Question 1: Add Two Numbers
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

```python
# Question 2: Square of a Number
def square(a):
    return a*a
    
result = square(6)
print(result)
```

```python
# Question 3: Check Even or Odd
def check(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check(10))
print(check(9))
```

```python
# Question 4: Find the Largest of Two Numbers
def largest_of_two(a,b):
    if a > b:
        return "A is largest",a
    else:
        return "B is largest",b
    
print(largest_of_two(12,34))
print(largest_of_two(10,7))
```

```python
# Question 5: Sum from 1 to N
def summ(n):
    total = 0

    for i in range(1,n+1):
        total = total + i
    return total

print(summ(5))
```

```python
# Question 6: Factorial of a Number
def factorial(n):
    if n < 0:
        return "Not for negative number"
    fact = 1
    for i in range(1,n+1):
        fact = fact * i
    return fact

print(factorial(5))
```

```python
# Question 7: Count the Digits
def count_digit(n):
    n = abs(n)
    if n == 0:
        return n
    count = 0
    while n > 0:
        n = n // 10
        count = count + 1
    return count

print(count_digit(0))
```

```python
# Question 8: Reverse a Number
def reverse_number(n):
    reverse = 0
    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10
    return reverse

print(reverse_number(123456789))
```

```python
# Question 9: Check a Prime Number
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2,n):
        if n % 2 == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(4))
```

```python
# Question 10: Find the Sum of Digits Using a Function
def sum_digit(n):
    total = 0
    while n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10
    return total

print(sum_digit(1234))
```
