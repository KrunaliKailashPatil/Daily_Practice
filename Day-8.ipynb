```python
# Question 1. Calculate Net Salary
def net_salary(salary):
    Allowance = salary * 20 
    Tax = salary * 10
    NetSalary = salary + Allowance - Tax
    return NetSalary

print(net_salary(30000))
```

```python
# Question 2. Calculate Profit or Loss
def profit_loss(cost,selling_price):
    if selling_price > cost:
        return "Profit=",selling_price - cost
    elif cost > selling_price:
        return "Loss=",cost - selling_price
    else:
        return "No Profit"
    
print(profit_loss(6000,3000))
```

```python
# Question 3. Calculate BMI
def bmi(weight,height):
    calculate_bmi = weight / (height * height)
    return round(calculate_bmi,2)

print(bmi(60,1.5))
```

```python
# Question 4. Calculate the Final Price After GST
def GST(price):
    calculate_gst = price * 18 / 100
    return price + calculate_gst

print(GST(1000))
```

```python
# Question 5. Count Words in a Sentence
def count_word(sentence):
    word = sentence.split()
    return len(word)

print(count_word("Python is a easy to learn"))
```

```python
# Question 6. Find Words Starting With a Specific Letter
def start_word(words,letter):
    result = []
    for word in words:
        if word.lower().startswith(letter.lower()):
            result.append(word)
    return result

print(start_word(["Python","Power BI","Excel"],"P"))
```

```python
# Question 7. Find the Difference Between Maximum and Minimum
def value(numbers):
    differnce = max(numbers) - min(numbers)
    return differnce

print(value([2000,3000,4000,5000]))
```

```python
# Question 8. Replace Negative Numbers With Zero
def replace(numbers):
    result = []
    for num in numbers:
        if num < 0:
            result.append(0)
        else:
            result.append(num)
    return result

print(replace([100,-2,-20,300,-3,-5]))
```

```python
# Question 9. Find Products Below a Stock Limit
def low_stock(stock):
    result = []

    for quantity in stock:
        if quantity < 10:
            result.append(quantity)
    return result

print(low_stock([12,14,10,25,9]))
```

```python
# Question 11. Find the Difference Between Two Lists
def two_list(list1,list2):
    result = []

    for item in list1:
        if item not in list2:
            result.append(item)
    return result

print(two_list([10,20,30,40],[30,40,50,60]))
```
