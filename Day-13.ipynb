```python
# Question 17: Find the Largest Number Without max()
number = [12,45,34,24,22,87,99,54,32,54]
largest = number[0]
for num in number:
    if num >= largest:
        largest = num
print(largest)
```

```python
# Question 18: Count Even Numbers
number = [12,45,34,24,22,87,99,54,32,54]
count = 0
for num in number:
    if num % 2 == 0:
        count = count + 1
print(count)
```

```python
# Question 19: Remove Duplicate Elements
number = [12,45,34,24,22,87,99,54,32,54,12,45,24,33]
unique = []
for num in number:
    if num not in unique:
        unique.append(num)
print(unique)
```

```python
# Question 20: Count Positive Numbers in a List
number = [12,45,34,24,22,87,99,54,32,54,-2,-4,-5,-7,-6,-3,-7]
positive = 0
for num in number:
    if num > 0:
        positive = positive + 1
print(positive)
```
