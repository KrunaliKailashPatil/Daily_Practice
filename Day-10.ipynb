```python
# Question 1. Calculate the Sum of Even Digits
def sum_even(number):
    total = 0
    while number > 0:
        digit = number % 10

        if number % 2 == 0:
            total = total + digit

        number = number // 10
    return total
print(sum_even(123456789))
```

```python
# Question 2. Calculate the Difference Between Odd and Even Counts
listt = [1,2,3,4,5,6,7]

def counts(listt):
    even = 0
    odd = 0

    for num in listt:
        if num % 2 == 0:
            even = even + 1
        else:
            odd = odd + 1
    return odd - even
print(counts(listt))
```

```python
# Question 3. Calculate the Final Price After a Fixed Discount
def discount(price):
    final = price - 200

    if final < 0:
        return 0
    else:
        return final
print(discount(900))
```

```python
# Question 4. Calculate the Average of Positive Numbers Only
listt=[-5, 10, 20, 0, 30]

def avg_positive(listt):
    total = 0
    count = 0

    for num in listt:
        if num > 0:
            total += num
            count += 1
    if count == 0:
        return 0
    return total / count

print(avg_positive(listt))
        
```

```python
# Question 5. Calculate the Percentage of Target Achieved
def target_archieved(actual,target):
    if target <= 0:
        return 0
    else:
        return (actual/target) * 100

print(target_archieved(60000,80000))
```

```python
# Question 6. Remove Empty Strings From a List
def remove_empty(words):
    result = []

    for word in words:
        if word != "":
            result.append(word)
    return result
print(remove_empty(["Data Sceinec","","AI"]))
```

```python
# Question 7. Find Words Longer Than a Given Length
def longer(words,length):
    result = []
    for word in words:
        if len(word) > length:
            result.append(word)
    return result
print(longer(["SQL","PYTHON","C++","JAVA"],4))
```

```python
# Question 8. Calculate Total Revenue From Sales Records
def total_revenue(records):
    total = 0

    for quantity,price in records:
        total += quantity * price

    return total
print(total_revenue([(2, 100), (3, 50), (1, 200)]))
```

```python
# Question 9. Find Customers With Missing Email Addresses
data = [("Neha","neha@gmail.com"),
        ("Krunali","Krunali@gmail.com"),
        ("Asha","")]

def missing_email(data):
    result = []

    for name,email in data:
        if email == "":
            result.append(name)
            
    return result
print(missing_email(data))
```

```python
# Question 10. Find the Highest Sales Day
data = [("Sunday",10000),
        ("Monday",40000),
        ("Tuesday",70000)]

def highest_sales_day(data):
    highest_day = ""
    highest_sales = 0

    for day,sales in data:
        if sales > highest_sales:
            highest_sales = sales
            highest_day = day
    return highest_day

print(highest_sales_day(data))
```
