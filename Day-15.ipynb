```python
# Question 1: Find the Third-Largest Unique Number
numbers = [10, 50, 30, 50, 40, 20]
unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

unique.sort(reverse=True)
print(unique[2])
```

```python
# Question 2: Count How Many Times Each Word Appears
words = ["apple", "banana", "apple", "orange", "banana", "apple"]
unique = []

for word in words:
    if word not in unique:
        unique.append(word)

for word in unique:
    print(word, "=", words.count(word))
```

```python
# Question 3: Find the Difference Between Adjacent Numbers
numbers = [10, 25, 40, 60]
differences = []

for i in range(1, len(numbers)):
    differences.append(numbers[i] - numbers[i - 1])
print(differences)
```

```python
# Question 4: Remove Empty Strings from a List
items = ["Python", "", "SQL", "", "Excel", "Power BI"]
result = []

for item in items:
    if item != "":
        result.append(item)
print(result)
```

```python
# Question 5: Find the Smallest Number in a List
numbers = [25, 10, 40, 5, 30]
smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num
print(smallest)
```

```python
# Question 6: Move Negative Numbers to the Beginning
numbers = [10, -5, 20, -2, 30, -8]
negative = []
positive = []

for num in numbers:
    if num < 0:
        negative.append(num)
    else:
        positive.append(num)
result = negative + positive
print(result)
```

```python
# Question 7: Find the Longest Word
words = ["SQL", "Python", "Analytics", "Excel", "Dashboard"]
longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)
```

```python
# Question 8: Replace Each Number with Its Square
numbers = [2, 4, 6, 8, 10]
squares = []

for num in numbers:
    squares.append(num ** 2)
print(squares)
```

```python
# Question 9: Find the Index of the First Repeated Value
numbers = [10, 20, 30, 20, 40]
seen = []

for i in range(len(numbers)):
    if numbers[i] in seen:
        print(i)
        break
    seen.append(numbers[i])
```

```python
# Question 10: Calculate Total Sales Above a Target
sales = [500, 1500, 2000, 800, 3000, 900]
total = 0

for amount in sales:
    if amount > 1000:
        total += amount
print(total)
```
