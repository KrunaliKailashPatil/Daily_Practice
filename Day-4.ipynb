```python
# 1. Check Leap Year
year = int(input("Enter a year:"))
if year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

```python
# 2. Factorial of a Number
num=int(input("Enter a number:"))
fact = 1
for i in range (1,num+1):
    fact = fact * i
print(fact)
```

```python
# 3. Reverse a Number
num=int(input("Enter a number:"))
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
print(rev)
```

```python
# 5. Count Digits
num=int(input("Enter a number:"))
count = 0
while num > 0:
    num = num // 10
    count = count + 1
print(count)
```

```python
# 6. Sum of Digits
num=int(input("Enter a number:"))
total = 0
while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10
print(total)
```

```python
# 8. Armstrong Number
num=int(input("Enter a number:"))
original = num
total = 0
digi = len(str(num))

while num > 0:
    digit = num % 10
    total = total + digit ** digi
    num = num // 10
if total==original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
    
```

```python
# 9. Find Largest Number in a List
listt=[23,54,6,43,66,224,88,98]
largest=listt[0]
for i in listt:
    if i > largest:
        largest = i
print(largest)
```

```python
# 10. Find Smallest Number in a List
listt=[23,54,6,43,66,224,88,98]
largest=listt[0]
for i in listt:
    if i < largest:
        largest = i
print(largest)
```
