```python
# Question 1: Calculate Compound Interest
compound_intrest = lambda p,r,t : p * (1+r/100)**t - p
print(round(compound_intrest(10000,10,2),2))
```

```python
# Question 2: Calculate the Hypotenuse of a Right Triangle
import math 
hypothenuse = lambda a,b : math.sqrt(a**2+b**2)
print(hypothenuse(3,4))
```

```python
# Question 3: Calculate the Roots' Product
root_product = lambda a,b,c : c/a
print(root_product(2,5,3))
```

```python
# Question 4: Calculate the Arithmetic Progression Sum
arithmetic = lambda a,d,n : n * (2*a+(n-1)*d)/2
print(arithmetic(2,3,5))
```

```python
# Question 5: Calculate the Geometric Progression Sum
geometric = lambda a,r,n : a * ((r**n)-1) / (r-1) if r != 1 else a*n
print(geometric(2,3,4))
```

```python
# Question 6: Calculate the Mean Absolute Deviation
data = [12,34,54,23]
mean = sum(data) / len(data)
mad = lambda values,avg:sum(abs(x-avg)for x in values) / len(values)
print(mad(data,mean))
```

```python
# Question 7: Calculate the Weighted Average
marks = [20,40,60]
weights = [3,4,5]
avg = lambda marks,weights : sum(m * w for m, w in zip(marks, weights)) / sum(weights)
print(round(avg(marks,weights)))
```

```python
# Question 8: Calculate the Percentage Change
sale = lambda old,new : ((new - old) / old) * 100
print(sale(8000,10000))
```

```python
# Question 9: Calculate Profit Margin
margin = lambda revenue,cost : ((revenue-cost)/revenue) * 100
print(margin(50000,35000))
```

```python
# Question 10: Calculate the Interquartile Range
interquartile = lambda q1,q3 : q3-q1
print(interquartile(20,60))
```
