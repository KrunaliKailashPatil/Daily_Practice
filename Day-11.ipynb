```python
# Question 1: Square of a Number
square = lambda a:a*a
print(square(5))
```

```python
# Question 2: Add Two Numbers
add = lambda a,b:a+b
print(add(10,20))
```

```python
# Question 3: Check Even or Odd
check = lambda a: "Even" if a % 2 == 0 else "Odd"
print(check(19))
```

```python
# Question 4: Find the Greater Number
greater = lambda a,b : a if a > b else b
print(greater(20,40))
```

```python
# Question 5: Convert Celsius to Fahrenheit
fahrenheit = lambda c : (c * 9/5) + 32
print(fahrenheit(30))
```

```python
# Question 6: Calculate Discounted Price
discount = lambda price:price - (price * 10/100)
print(discount(1000))
```

```python
# Question 7: Calculate Total Sales
total_sales = lambda quantity,price : quantity*price
print(total_sales(10,1000))
```

```python
# Question 8: Calculate Average Marks
avergae = lambda a,b,c : (a+b+c)/3
print(int(avergae(12,23,34)))
```

```python
# Question 9: Calculate Employee Bonus
bonus = lambda salary : salary * (5/100)
print(bonus(3000))
```

```python
# Question 10: Categorize Customer Spending
category = lambda money : "High" if money > 10000 else "Low"
print(category(6000))
```

```python
# Question 11: Calculate the Cube of a Number
cube = lambda n:n**3
print(cube(4))
```

```python
# Question 12: Calculate Simple Interest
intrest = lambda p,r,t: (p*r*t)/100
print(intrest(10000,5,2))
```

```python
# Question 13: Calculate the Area of a Circle
circle = lambda r : 3.14 * r ** 2
print(circle(5))
```

```python
# Question 14: Calculate Percentage
percentage = lambda obtained , total=500: (obtained/total) * 100
print(percentage(344))
```

```python
# Question 15: Find the Remainder
remainder = lambda a,b : a % b
print(remainder(17,5))
```
